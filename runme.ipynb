```python
from music21 import converter, instrument, note, chord, stream, midi
from pathlib import Path

class MidiToSheetMusic:
    def __init__(self, midi_file_path='recording.mid'):
        self.midi_file_path = Path(midi_file_path)
        self.midi_data = None
        self.parsed_notes = []

    def load_midi(self):
        """Load and parse the MIDI file."""
        self.midi_data = converter.parse(self.midi_file_path)
        self.parse_midi_notes()

    def parse_midi_notes(self):
        """Parse MIDI file to extract notes with correct durations."""
        midi_stream = midi.MidiFile()
        midi_stream.open(self.midi_file_path)
        midi_stream.read()
        midi_stream.close()

        notes_dict = {}
        for track in midi_stream.tracks:
            for event in track.events:
                if event.isNoteOn():
                    notes_dict[event.pitch] = event.time
                elif event.isNoteOff():
                    start_time = notes_dict.pop(event.pitch, None)
                    if start_time is not None:
                        duration = event.time - start_time
                        self.parsed_notes.append((event.pitch, duration))


    # def save_sheet_music(self, file_format='musicxml', output_path='output_sheet_music.xml'):
    #     """
    #     Save the sheet music to a file.
        
    #     Args:
    #         file_format: The format to save the sheet music ('musicxml', 'lily.pdf', etc.).
    #         output_path: The output file path.
    #     """
    #     if self.midi_data is not None:
    #         self.midi_data.write(file_format, output_path)
    #     else:
    #         raise Exception("MIDI file is not loaded. Please load the MIDI file first.")

    def print_notes_and_chords(self):
        """Print all notes and chords from the MIDI file."""
        if self.midi_data is not None:
            for pitch, duration in self.parsed_notes:
                note_name = note.Note(pitch).name
                print(f"Note: {note_name} Duration: {duration}")
        else:
            raise Exception("MIDI file is not loaded. Please load the MIDI file first.")

# Example usage:
# midi_to_sheet = MidiToSheetMusic('path_to_your_midi_file.mid')
# midi_to_sheet.load_midi()
# midi_to_sheet.show_sheet_music()
# midi_to_sheet.save_sheet_music(file_format='lily.pdf', output_path='output_sheet_music.pdf')
# midi_to_sheet.print_notes_and_chords()
```
